```python
#write your code here

import numpy as np
from numpy.linalg import LinAlgError

def is_basis_and_decompose(vectors, b):
    """
    vectors: список векторів (кожен вектор — список або np.array)
    b: вектор, який потрібно розкласти
    """

    A = np.column_stack(vectors)  # матриця, у якій вектори — стовпці
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    print("Матриця базису A:")
    print(A)
    print("\nВектор b:")
    print(b)

    try:
        # перевіримо, чи матриця невироджена (тобто det ≠ 0)
        det = np.linalg.det(A)
        print(f"\nВизначник матриці A: {det}")

        if np.isclose(det, 0):
            print("Вектори не утворюють базис: визначник дорівнює нулю.")
            return

        # Розв'язуємо A * x = b
        x = np.linalg.solve(A, b)
        print("\nРозклад вектора b за базисом:")
        for i, coeff in enumerate(x):
            print(f"  {coeff:.4f} * a_{i+1}")
        return x

    except LinAlgError as e:
        print("Помилка при обчисленні:", e)

# Приклад використання
a1 = [1, 0, 0]
a2 = [0, 1, 0]
a3 = [0, 0, 1]
b = [2, -3, 5]

vectors = [a1, a2, a3]
is_basis_and_decompose(vectors, b)
```

Протестуйте ваші функції та продемонструйте їхню роботу.

```python
import numpy as np
from sympy import Matrix
from sympy import gcd  # використовуємо gcd із sympy

# 1. Алфавіт і відповідність символів елементам поля Z_29
alphabet = [chr(i) for i in range(65, 91)] + [' ', '.', ',']
char_to_num = {char: i for i, char in enumerate(alphabet)}
num_to_char = {i: char for char, i in char_to_num.items()}

# 2. Допоміжна функція: розбиття на блоки фіксованої довжини
def split_text(text, n):
    text = text.upper()
    padding = (n - len(text) % n) % n
    text += ' ' * padding  # доповнюємо пробілами
    return [text[i:i+n] for i in range(0, len(text), n)]

# 3. Текст <-> Вектор
def text_to_vector_block(block):
    return np.array([char_to_num[c] for c in block])

def vector_block_to_text(vec):
    return ''.join([num_to_char[int(x) % 29] for x in vec])

# 4. Перевірка на обертовність та знаходження оберненої матриці
def is_invertible_mod(matrix, mod=29):
    det = round(Matrix(matrix.tolist()).det()) % mod
    return det != 0 and gcd(det, mod) == 1

def inverse_mod_matrix(matrix, mod=29):
    sympy_matrix = Matrix(matrix.tolist())
    inv_matrix = sympy_matrix.inv_mod(mod)
    return np.array(inv_matrix.tolist()).astype(int)

# 5. Функція шифрування
def encrypt(text, key_matrix):
    n = key_matrix.shape[0]
    blocks = split_text(text, n)
    encrypted_blocks = []

    for block in blocks:
        vec = text_to_vector_block(block).reshape(-1, 1)
        enc_vec = key_matrix.dot(vec) % 29
        encrypted_blocks.append(enc_vec.flatten())

    return np.array(encrypted_blocks)

# 6. Функція дешифрування
def decrypt(encrypted_blocks, key_matrix):
    inv_key = inverse_mod_matrix(key_matrix, 29)
    decrypted_text = ""

    for vec in encrypted_blocks:
        vec = vec.reshape(-1, 1)
        dec_vec = inv_key.dot(vec) % 29
        decrypted_text += vector_block_to_text(dec_vec.flatten())

    return decrypted_text.strip()

# 7. Приклад роботи
key = np.array([[3, 10], [20, 9]])  # Простий ключ 2x2

if is_invertible_mod(key):
    msg = "HELLO, Alice from Bob"
    print("Оригінальне повідомлення:", msg)

    encrypted = encrypt(msg, key)
    print("Зашифроване повідомлення (вектори):")
    for vec in encrypted:
        print(vec)

    decrypted = decrypt(encrypted, key)
    print("Розшифроване повідомлення:", decrypted)
else:
    print("Матриця не є обертовною в Z_29.")
```

### Кількість можливих обернених матриць-ключів над полем $\mathbb{Z}_{29}$

Шифр Гілла використовує як ключ квадратну матрицю розміру $n \times n$, елементи якої належать полю $\mathbb{Z}_{29}$. Щоб матриця була придатною для шифрування, вона повинна бути **оберненою** (невиродженою) в цьому полі.

Загальна формула для обчислення кількості обернених $n \times n$ матриць над полем з  $q$ елементів:

$$
|K| = q^{n^2} \cdot \prod_{i=1}^{n} \left(1 - \frac{1}{q^i} \right)
$$

Підставляючи \( q = 29 \), отримаємо:

$$
|K| = 29^{n^2} \cdot \left(1 - \frac{1}{29}\right) \left(1 - \frac{1}{29^2}\right) \cdots \left(1 - \frac{1}{29^n}\right)
$$



### Приклад: $n = 3$

Підставимо $n = 3$ у формулу:

$$
|GL(3, \mathbb{Z}_{29})| = (29^3 - 1)(29^3 - 29)(29^3 - 29^2)
$$

Обчислимо:

- $29^3 = 24389$
- $24389 - 1 = 24388$
- $24389 - 29 = 24360$
- $24389 - 841 = 23548$

$$
|K| = 24388 \times 24360 \times 23548 \approx 1.4 \times 10^{13}
$$


## Висновок

- Навіть для $n = 3$ кількість можливих ключів — понад $10^{13}$.
- Для $n = 4$ — це вже більше $10^{23}$.
- Це означає, що **груба сила** є **непрактичною** для будь-якого $n \geq 3$.

## 3️⃣ Чи можливо застосувати частотний аналіз до шифру Гілла?

## Ідея методу

Метод **частотного аналізу** базується на тому, що в будь-якій мові певні символи або комбінації символів зустрічаються **значно частіше**, ніж інші.

У шифрі Гілла текст шифрується **не посимвольно**, а **по блоках з $n$ символів**, які інтерпретуються як вектори в просторі $\mathbb{Z}_{29}^n$.

Припустимо, що відкритий текст розбитий на $s$ блоків довжини $n$:  
$$
u_1, u_2, \dots, u_s
$$
Зашифровані блоки:
$$
Ku_1, Ku_2, \dots, Ku_s
$$
де $K$ — матриця-ключ розміру $n \times n$ над $\mathbb{Z}_{29}$.

Тобто **один зашифрований символ залежить одразу від усього блоку з $n$ символів відкритого тексту**.


## Наслідки для частотного аналізу

Це означає:

- **Немає прямого відображення** між відкритими символами та зашифрованими — один символ шифротексту **не пов’язаний** з одним символом відкритого тексту.
- **Частоти окремих символів не зберігаються**.  
  Усі блоки відкритого тексту (навіть якщо це однакові слова) при різних $K$ можуть дати **абсолютно різні шифровані блоки**.



## Але чи можливо провести аналіз за блоками?

Якщо довжина блоку $n$ є невеликою (наприклад, $n = 2$ або $n = 3$), то:

- Можна здійснити **частотний аналіз для блоків з $n$ символів**.
- Наприклад, визначити, які блоки найчастіше зустрічаються в шифротексті.
- І припустити, що вони відповідають найпоширенішим словам або комбінаціям символів (наприклад: "на", "по", "the", "ing").

Проте кількість можливих блоків швидко зростає:

\( \text{Кількість блоків довжини } n = 29^n \)

Наприклад:

- $n = 2$: $29^2 = 841$ блок
- $n = 3$: $29^3 = 24\,389$ блоків

Чим більший $n$, тим **менше шансів**, що певний блок часто повторюватиметься — отже частотний аналіз **втрачає ефективність**.



## Висновок

- **Частотний аналіз на рівні символів — не працює** у шифрі Гілла, бо символи повністю "розмазуються" у блоках.
- **Частотний аналіз блоків** можливий лише при **малому $n$**, або якщо дані мають **повторювану структуру** (наприклад, однакові короткі слова).
- Тому **не рекомендується використовувати шифр Гілла для структурованих даних**, де часто повторюються однакові блоки (наприклад, назви, префікси, тощо).
- При **випадковому ключі** і достатньо великому $n$ (від $3$ і вище) — **частотний аналіз є неефективним засобом злому**.

## Метод «відомого відкритого тексту» для зламу шифра Гілла

## Ідея методу

У багатьох випадках ворог (Єва) може **знати або припускати**, який саме відкритий текст (або його частина) відповідає певному зашифрованому фрагменту. Це і є суть **методу відомого відкритого тексту**.

Зокрема, вона має:

- $u_1, \dots, u_s$ — фрагменти **відомого відкритого тексту**, представлені як вектори довжини $n$
- $v_1, \dots, v_s$ — відповідні **зашифровані вектори**, тобто фрагменти шифротексту

Модель шифрування для кожного блоку:
$$
v_i = K u_i, \quad i = 1, 2, \dots, s
$$
де $K$ — **матриця-ключ** розміру $n \times n$ над полем $\mathbb{Z}_{29}$.



## Як знайти матрицю $K$

Запишемо всі відомі пари $u_i, v_i$ у вигляді матриць:

- Нехай $U$ — матриця, утворена з векторів $u_i$ як **стовпців**:
$$
U = \begin{bmatrix} u_1 & u_2 & \dots & u_n \end{bmatrix}
$$
- Нехай $V = [v_1 \ v_2 \ \dots \ v_n]$ — відповідні зашифровані вектори

Маємо:
$$
V = K U
\quad \Rightarrow \quad
K = V U^{-1}
$$
тобто, щоб знайти $K$, Єва повинна:

1. Зібрати **$n$ лінійно незалежних векторів $u_i$**, які утворюють **невироджену** (обернену) матрицю $U$
2. Мати відповідні вектори $v_i$
3. Обчислити обернену матрицю $U^{-1}$ над полем $\mathbb{Z}_{29}$
4. Знайти ключ:  
$$
K = V U^{-1} \mod 29
$$

Щоб відновити матрицю $K$ розміру $n \times n$, потрібно **щонайменше $n$ рівнянь** (векторів $u_i$), тобто:

- **$n$ відомих пар**: $u_i \leftrightarrow v_i$
- **$U$ має бути оберненою**, тобто її стовпці повинні бути **лінійно незалежними** у $\mathbb{Z}_{29}^n$
